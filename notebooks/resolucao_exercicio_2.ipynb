## Text Summarization Pipeline Using LangChain

### 1.Import libraries

```python
import os
import sys
import warnings


BASE_DIR = os.path.abspath("..")

sys.path.append(BASE_DIR)

from src import split_text, summarize, load_txt_document, load_pdf_document, size_file, print_chunks, TextCleaner, validate_summary

#warnings.filterwarnings("ignore", category=FutureWarning)
```

### 2.Read file and text cleaning

```python
def read_files(file_path):
    if file_path.endswith(".pdf"):
        documents = load_pdf_document(file_path)
        size_file(file_path)
    else:
        documents = load_txt_document(file_path)
        size_file(file_path)


    # clean documents
    for doc in documents:

        cleaner = TextCleaner(doc.page_content)

        cleaner.remove_extra_spaces()
        cleaner.remove_special_characters()
        cleaner.to_lowercase()

        doc.page_content = cleaner.get_clean_text()
    return documents
```

### 3.Split in chunks with RecursiveCharacterTextSplitter

```python
def split_chunks(documents):

    chunks = split_text(documents)

    #Chunks detail
    print_chunks(chunks)

    return chunks
```

### 4. Summarize with LLM + map_reduce

```python
def summarize_chunks(chunks, mode, prompt):
    summary = summarize(chunks, mode, prompt)
    print(summary)
    validation = validate_summary(summary, chunks)    
    print(validation)
```

### 5-Validations

#### 1º Case:

```python
path = os.path.join(BASE_DIR, "data", "easy_recipes.pdf")

print('\nREADING FILE...')
documents = read_files(path)

print('\n\nSPLITTING CHUNKS...')
chunks= split_chunks(documents)

print('\n\nSUMMARIZING..WAIT UNTIL PROCESSING IS FINISHED!!\n')
summary_result = summarize_chunks(chunks, mode="short", prompt="Summarize this document into a high-level overview.")
```
