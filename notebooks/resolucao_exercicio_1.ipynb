## Python Fundamentals — Mandatory Text Processing Script

### 1.Importação de funções e bibliotecas

```python
import re
import os
import sys

sys.path.append(os.path.abspath(".."))

from src import read_text_file, TextCleaner, extract_emails, validate_employee_ids, word_frequency
```

### 2.Importação de arquivo

```python
BASE_DIR = "/app"
sample_path = os.path.join(BASE_DIR, "data", "sample.txt")

text = read_text_file(sample_path)

print(text)
```

### 3.Limpar texto

```python
cleaner = TextCleaner(text)

cleaner.remove_extra_spaces()
cleaner.remove_special_characters()
cleaner.to_lowercase()

clean_text = cleaner.get_clean_text()

print(clean_text)
```

### 4.Extrair emails

```python
emails = extract_emails(clean_text)

print("Extracted Emails:")
print(emails)
```

### 5.Validar Ids

```python
employee_ids = re.findall(r'\b[A-Z0-9]{3}-\d{4}\b', text)

valid_ids, invalid_ids = validate_employee_ids(employee_ids)

print("Valid IDs:")
print(valid_ids)

print("\nInvalid IDs:")
print(invalid_ids)
```

### 6.Frequência das palavras

```python
frequency = word_frequency(clean_text)

print("Word Frequency:")
print(frequency)
```
